# Exploratory Data Analysis: Palmer Penguins Dataset

## Overview
This project explores the Palmer Penguins dataset to analyze species differences, physical measurements, and patterns across islands. The goal is to uncover insights using descriptive statistics and visualizations.


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set(style="whitegrid")
```

```python
df_clean = df.dropna().copy()
df_clean.shape
```

```python
df_clean.hist(figsize=(10,6))
plt.tight_layout()
plt.show()
```

```python
sns.boxplot(data=df_clean, x="species", y="body_mass_g")
plt.show()
```

```python
sns.scatterplot(
    data=df_clean,
    x="bill_length_mm",
    y="flipper_length_mm",
    hue="species"
)
plt.show()
```

## Key Insights

- Gentoo penguins have the highest average body mass.
- There is a positive relationship between bill length and flipper length.
- Species show distinct clustering in physical measurements.
- Removing missing values improved analysis reliability.
